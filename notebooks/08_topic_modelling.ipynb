```python
import sys
import os
sys.path.append(os.path.abspath(".."))
```

```python
with open("../data/processed/cleaned_transcript.txt", "r", encoding="utf-8") as f:
    cleaned_text = f.read()

print(cleaned_text[:300])
```

```python
from src.topic_modeling import TopicModeler

modeler = TopicModeler(num_topics=2, num_keywords=5)

results = modeler.structured_output(cleaned_text)

print("Top Keywords:")
print(results["keywords"])

print("\nTopics:")
print(results["topics"])
```

```python
import json

output_path = "../data/processed/topic_results.json"

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(results, f, indent=4)

print("Topic modeling results saved ✅")
```
