```python
import os

TRANSCRIPT_PATH = "../data/processed/transcript.txt"

with open(TRANSCRIPT_PATH, "r", encoding="utf-8") as f:
    raw_text = f.read()

print("Transcript Loaded ✅")
print(raw_text[:500])  # show first 500 chars
```

```python
import re

def basic_cleaning(text):
    text = text.lower()
    text = re.sub(r"\s+", " ", text)  # remove extra spaces
    text = text.strip()
    return text

clean_text = basic_cleaning(raw_text)

print("After basic cleaning:\n")
print(clean_text[:500])
```

```python
def remove_fillers(text):
    fillers = ["um", "uh", "you know", "like", "basically", "actually"]
    
    for filler in fillers:
        text = text.replace(filler, "")
    
    return text

clean_text = remove_fillers(clean_text)

print("After removing fillers:\n")
print(clean_text[:500])
```

```python
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(clean_text)

print("Number of sentences:", len(sentences))
print("\nFirst 3 sentences:\n")
print(sentences[:3])
```

```python
from nltk.tokenize import word_tokenize

words = word_tokenize(clean_text)

print("Total words:", len(words))
print("\nFirst 20 words:\n")
print(words[:20])
```

```python
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

filtered_words = [word for word in words if word.isalpha() and word not in stop_words]

print("Total filtered words:", len(filtered_words))
print("\nFirst 20 filtered words:\n")
print(filtered_words[:20])
```

```python
print("Original word count:", len(words))
print("After stopword removal:", len(filtered_words))
print("Number of sentences:", len(sentences))
```

```python
CLEANED_PATH = "../data/processed/cleaned_transcript.txt"

with open(CLEANED_PATH, "w", encoding="utf-8") as f:
    f.write(" ".join(filtered_words))

print("Cleaned transcript saved ✅")
```
