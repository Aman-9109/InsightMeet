```python
import os
import librosa
import matplotlib.pyplot as plt

AUDIO_PATH = "../data/raw/sample_meeting.wav"

audio, sr = librosa.load(AUDIO_PATH, sr=None)

print("Sample Rate:", sr)
print("Audio Length (seconds):", len(audio)/sr)

plt.figure()
plt.plot(audio)
plt.title("Audio Waveform")
plt.show()
```
