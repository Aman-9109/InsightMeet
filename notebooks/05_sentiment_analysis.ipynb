```python
import sys
import os

sys.path.append(os.path.abspath(".."))
```

```python
import json

with open("../data/processed/chunks.json", "r", encoding="utf-8") as f:
    chunks = json.load(f)

print("Chunks loaded:", len(chunks))
print("\nFirst chunk preview:\n")
print(chunks[0][:300])
```

```python
from src.sentiment import SentimentAnalyzer

analyzer = SentimentAnalyzer()

chunk_results = analyzer.analyze_chunks(chunks)

print("Chunk-level sentiment:\n")
print(chunk_results)
```

```python
overall = analyzer.aggregate_sentiment(chunk_results)

print("\nOverall Meeting Sentiment:")
print(overall)
```

```python
output_path = "../data/processed/sentiment_results.json"

with open(output_path, "w", encoding="utf-8") as f:
    json.dump({
        "chunk_results": chunk_results,
        "overall": overall
    }, f, indent=4)

print("Sentiment results saved ✅")
```
