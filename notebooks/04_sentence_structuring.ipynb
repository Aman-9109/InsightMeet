```python
import sys
import os

sys.path.append(os.path.abspath(".."))
```

```python
with open("../data/processed/cleaned_transcript.txt", "r", encoding="utf-8") as f:
    cleaned_text = f.read()

print("Loaded cleaned transcript ✅")
print(cleaned_text[:500])
```

```python
from src.chunking import TextChunker

chunker = TextChunker(max_words=100)

sentences = chunker.split_sentences(cleaned_text)

print("Total sentences:", len(sentences))
print("\nFirst 3 sentences:\n")
print(sentences[:3])
```

```python
chunks = chunker.create_chunks(cleaned_text)

print("Total chunks created:", len(chunks))
print("\nFirst chunk preview:\n")
print(chunks[0][:500])
```

```python
import json

CHUNK_PATH = "../data/processed/chunks.json"

with open(CHUNK_PATH, "w", encoding="utf-8") as f:
    json.dump(chunks, f, indent=4)

print("Chunks saved successfully ✅")
```
