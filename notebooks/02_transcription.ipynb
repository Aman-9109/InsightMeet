```python
import whisper

print("Whisper version loaded successfully ✅")
```

```python
model = whisper.load_model("base")
print("Model loaded successfully ✅")
```

```python
import os

AUDIO_PATH = "../data/raw/sample_meeting.wav"

print("Absolute Path:")
print(os.path.abspath(AUDIO_PATH))

print("File Exists?")
print(os.path.exists(AUDIO_PATH))
```

```python
result = model.transcribe(AUDIO_PATH)

print("Transcription Complete ✅")
print(result["text"])
```

```python
OUTPUT_PATH = "../data/processed/transcript.txt"

os.makedirs("../data/processed", exist_ok=True)

with open(OUTPUT_PATH, "w", encoding="utf-8") as f:
    f.write(result["text"])

print("Transcript saved at:")
print(os.path.abspath(OUTPUT_PATH))
```

```python
def transcribe_audio(audio_path, model):
    try:
        if not os.path.exists(audio_path):
            raise FileNotFoundError("Audio file not found.")

        result = model.transcribe(audio_path)
        return result["text"]

    except Exception as e:
        print("Error occurred:", e)
        return None
```

```python
text = transcribe_audio(AUDIO_PATH, model)
print(text)
```
